## Concept clé

Une architecture production n’est pas un appel direct au modèle. Elle encapsule modèle, outils, état, mémoire, sécurité, observabilité et rollback dans des frontières explicites.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
if (PROJECT_ROOT / "ai_platform").exists():
    sys.path.insert(0, str(PROJECT_ROOT))

from ai_platform.architecture import (
    build_support_ai_blueprint,
    ProductionArchitectureValidator,
    summarize_readiness,
    render_mermaid,
)

blueprint = build_support_ai_blueprint()
report = ProductionArchitectureValidator().validate(blueprint)
print(summarize_readiness(report))
```
