## Concepts

- Trace : opération complète.
- Span : sous-opération mesurée.
- Event : fait ponctuel.
- Metric : point numérique agrégable.
- Redaction : protection des données sensibles.

```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from mini_framework.observability import Observability, ObservabilityConfig
```

```python
obs = Observability(ObservabilityConfig(service_name="notebook_demo"))

with obs.start_trace("agent_run", {"email": "client@example.com"}) as trace:
    with obs.start_span("agent.plan", kind="agent"):
        obs.record_event("plan_created", "A simple plan was created")
    with obs.start_span("tool.search", kind="tool", attributes={"api_key": "sk-secret"}):
        obs.record_metric("tool.call.count", 1)

obs.export_json()
```
