```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from mini_framework.integration import build_support_framework
```

```python
framework = build_support_framework()
framework.memory.remember("student_user", "preferred_tone", "empathique")
result = framework.run(
    user_id="student_user",
    message="Bonjour, ma commande est cassée et je veux un remboursement.",
    workflow_name="support_resolution",
)
result.to_dict()
```

# Notes formateur

Le point important est que le framework peut produire un résultat utile même si une action sensible est bloquée.

## Corrigés attendus

- Un remboursement est une action sensible.
- Le framework ne doit pas l'exécuter sans approbation.
- L'agent n'appelle pas directement les fonctions.
- Le registre d'outils valide permissions et arguments.
- La trace doit expliquer ce qui a été fait ou bloqué.

```python
assert result.status == "completed"
assert "draft" in result.outputs
assert blocked_result.status == "blocked"
assert blocked_result.blocked_actions == ["issue_refund"]
```
